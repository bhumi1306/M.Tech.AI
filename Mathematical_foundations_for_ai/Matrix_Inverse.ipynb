```python
def transpose_matrix(matrix):
    return [list(row) for row in zip(*matrix)]
```

```python
def get_minor(matrix, row, col):
    return [
        r[:col] + r[col + 1:]
        for r in (matrix[:row] + matrix[row + 1:])
    ]
```

```python
def determinant(matrix):
    n = len(matrix)

    if n == 1:
        return matrix[0][0]

    if n == 2:
        return (
            matrix[0][0] * matrix[1][1]
            - matrix[0][1] * matrix[1][0]
        )

    det = 0

    for col in range(n):
        minor = get_minor(matrix, 0, col)

        cofactor = ((-1) ** col) * determinant(minor)

        det += matrix[0][col] * cofactor

    return det
```

```python
def cofactor_matrix(matrix):
    n = len(matrix)
    cofactors = []

    for row in range(n):
        cofactor_row = []

        for col in range(n):
            minor = get_minor(matrix, row, col)

            cofactor = (
                (-1) ** (row + col)
                * determinant(minor)
            )

            cofactor_row.append(cofactor)

        cofactors.append(cofactor_row)

    return cofactors
```

```python
def adjoint(matrix):
    cofactors = cofactor_matrix(matrix)

    return transpose_matrix(cofactors)
```

```python
def inverse(matrix):
    det = determinant(matrix)

    if det == 0:
        raise ValueError("Inverse does not exist because determinant is 0.")

    adj = adjoint(matrix)

    inverse_matrix = []

    for i in range(len(matrix)):
        row = []

        for j in range(len(matrix)):
            row.append(adj[i][j] / det)

        inverse_matrix.append(row)

    return inverse_matrix
```

```python
def print_matrix(matrix):
    print("[")
    
    for row in matrix:
        print("  " + "  ".join(f"{value:10.3f}" for value in row))
    
    print("]")
```

```python
n = int(input("Enter order of matrix: "))

matrix = []

print("Enter matrix elements separated by space:")

for i in range(n):
    row = list(map(float, input(f"Row {i + 1}: ").split()))

    if len(row) != n:
        raise ValueError(f"Enter exactly {n} elements.")

    matrix.append(row)
```

```python
print("\nMatrix:")
print_matrix(matrix)

det = determinant(matrix)

print("\nDeterminant:")
print(f"{det:.3f}")

print("\nCofactor Matrix:")
print_matrix(cofactor_matrix(matrix))

print("\nAdjoint:")
print_matrix(adjoint(matrix))

if det != 0:
    print("\nInverse:")
    print_matrix(inverse(matrix))
else:
    print("\nInverse does not exist.")
```
